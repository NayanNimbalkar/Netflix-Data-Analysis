```python
# Make a copy of the dataframe
genre_data = df.copy()

# Convert date_added to datetime
genre_data['date_added'] = pd.to_datetime(
    genre_data['date_added'],
    errors='coerce'
)

# Create year directly from date_added
genre_data['year_added'] = genre_data['date_added'].dt.year

# Split multiple genres into separate rows
genre_data['listed_in'] = (
    genre_data['listed_in']
    .fillna('')
    .str.split(', ')
)

genre_data = genre_data.explode('listed_in')

# Find the 5 most common genres
top_genres = (
    genre_data['listed_in']
    .value_counts()
    .head(5)
    .index
)

# Keep only the top 5 genres
genre_trend = genre_data[
    genre_data['listed_in'].isin(top_genres)
]

# Count each genre by year
genre_year = (
    genre_trend
    .groupby(['year_added', 'listed_in'])
    .size()
    .unstack(fill_value=0)
)


# ==========================================
# NETFLIX COLOR PALETTE
# ==========================================

colors = [
    "#E50914",  # Netflix Red
    "#B20710",  # Dark Red
    "#D85C63",  # Muted Red
    "#E87C83",  # Light Red
    "#8F1D2C"   # Deep Red
]


# ==========================================
# PLOT
# ==========================================

plt.figure(figsize=(12, 6))

for genre, color in zip(top_genres, colors):

    plt.plot(
        genre_year.index,
        genre_year[genre],
        marker='o',
        linewidth=2.5,
        markersize=5,
        color=color,
        label=genre
    )

plt.title(
    'Top Genre Addition Trends Over Time',
    fontsize=15,
    fontweight='bold'
)

plt.xlabel('Year Added')
plt.ylabel('Number of Titles')

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
